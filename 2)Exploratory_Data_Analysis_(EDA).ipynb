```python
import pandas as pd
data=pd.read_csv('/content/TItanic_dataset.csv')
```

```python
data.head()
```

```python
s=data.describe()
print(s)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.hist(data['Age'].dropna(),bins=20,edgecolor="red",color="skyblue")
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(6,4))
data['Survived'].value_counts().plot(kind='bar',color="green")
plt.title("Survival")
plt.xlabel("Class")
plt.ylabel("Count")
plt.xticks(rotation=1)
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.boxplot(data['Age'].dropna(),patch_artist=True,boxprops=dict(facecolor='black'))
plt.title('Box Plot Age')
plt.ylabel('Age')
plt.show()
```

```python
import seaborn as sns
```

```python
sns.pairplot(data[['Age','Fare','SibSp','Parch']].dropna())
plt.show()
```

```python
correlation_matrix=data[['Age','Fare','SibSp','Parch']].corr()
sns.heatmap(correlation_matrix,annot=True,cmap='coolwarm',cbar=True)
plt.title('Correlation Matrix')
plt.show()
```

```python
sns.countplot(x='Survived',data=data,palette='Set2')
plt.title('Survival Count')
plt.xlabel('Survived')
plt.ylabel('Count')
plt.show()
```

```python
sns.violinplot(x='Pclass',y='Age',data=data,palette='muted')
plt.title('Age Distribution by Passenger Class')
plt.ylabel('Age')
```
