```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data=pd.read_csv('/content/titanic_synthetic_418.csv')
data.head()
```

```python
data.info()
```

```python
data.describe()
```

```python
data.isnull().sum()
```

```python
data['Age'].fillna(data['Age'].mean(),inplace=True)
```

```python
print(data.isnull().sum())
```

```python
numerical_data = data.select_dtypes(include=['number'])
correlation_matrix = numerical_data.corr() # Fixed the typo here
plt.figure(figsize=(10, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f', cbar=True)
plt.title('Correlation Matrix')
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(x='Age',y='Fare',data=data,hue='Survived',palette='Set1')
plt.title('Age vs Fare:Analysis of Trend')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.histplot(data['Age'],bins=30,kde=True,color='red')
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.boxplot(x=data['Fare'],color='lightcoral')
plt.title('Fare Distribution')
plt.xlabel('Fare')
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.countplot(x='Pclass', data=data, palette='Set2')
plt.title('Passenger Class Distribution')
plt.xlabel('Pclass')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(10,8))
sns.countplot(x='Survived',data=data,palette='Set2')
plt.title('Survival Distribution ')
plt.xlabel('Survived')
plt.ylabel('Count')
plt.show()
```
